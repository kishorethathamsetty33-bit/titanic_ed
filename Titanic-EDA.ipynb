```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.drop("Cabin",axis=1,inplace=True)
```

```python
df.isnull().sum()
```

```python
df
```

```python
df["Survived"].value_counts().plot(kind="bar")
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
```

## Key Takeaways

- More passengers did not survive (0) than survived (1).
- The number of non-survivors is significantly higher than the number of survivors.
- The dataset shows that the survival rate was lower than the non-survival rate.

```python
df["Sex"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

## Key Takeaways

- Male passengers are more than female passengers.
- Female passengers represent a smaller portion of the dataset.
- The dataset is dominated by male passengers.

```python
df["Age"].plot(kind="hist", bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

## Key Takeaways

- Most passengers are between 20 and 40 years of age.
- There are fewer passengers in the higher age groups.
- The age distribution shows that young adults make up the majority of passengers.

```python
df["Pclass"].value_counts().sort_index().plot(kind="bar")

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

## Key Takeaways

- Most passengers belonged to Third Class (Pclass 3).
- First Class had fewer passengers than Third Class.
- Passenger distribution was not equal across the three classes.

```python
df["Fare"].plot(kind="hist", bins=20)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

## Key Takeaways

- Most passengers paid lower fares.
- A small number of passengers paid very high fares.
- The fare distribution contains a few outliers.
